### 4. Statistical Stationarity Testing (ADF Test) 
##### Stationarity (constant mean and variance over time) is a strict requirement for classical statistical models like ARIMA.

```python
from statsmodels.tsa.stattools import adfuller


# 1. Define the check_stationarity function FIRST
def check_stationarity(series, title='Time Series'):
    print(f'=== Augmented Dickey-Fuller (ADF) Testing: {title} ===')

    # Drop NA values generated by differencing before testing
    clean_series = series.dropna()

    result = adfuller(clean_series)
    adf_stat = result[0]
    p_value = result[1]

    print(f'ADF Statistic: {adf_stat:.4f}')
    print(f'p-value: {p_value:.4f}')
    print(result)

    if p_value < 0.05:
        print('Result: Series is STATIONARY (p < 0.05)\n')
    else:
        print(
            'Result: Series is NON-STATIONARY (p >= 0.05). Differencing'
            ' recommended.\n'
        )


# -------------------------------------------------------------------------
# 2. Resample to Continuous Daily Frequency
# -------------------------------------------------------------------------
df_daily = (
    Invoice_Date_Aggregate.set_index(Date_Col).resample('D').sum().reset_index()
)

# -------------------------------------------------------------------------
# 3. Call check_stationarity (Fixed Argument Syntax)
# -------------------------------------------------------------------------
# Corrected syntax: pass series as 1st argument, title string as 2nd argument
check_stationarity(df_daily[Target_Col], 'Continuous Daily Revenue')

# Create 1st Order Differencing
df_daily['Target_Diff_1'] = df_daily[Target_Col].diff()

# Test 1st Order Differenced Series
check_stationarity(df_daily['Target_Diff_1'], '1st Order Differenced Revenue')
```

```python
import matplotlib.pyplot as plt


# Set up figure
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8))

# 1. Autocorrelation Plot (Determines 'q' parameter for MA)
plot_acf(
    df_daily[Target_Col].dropna(),
    lags=31,
    ax=ax1,
    title='Autocorrelation Function (ACF)',
)

# 2. Partial Autocorrelation Plot (Determines 'p' parameter for AR)
plot_pacf(
    df_daily[Target_Col].dropna(),
    lags=31,
    ax=ax2,
    title='Partial Autocorrelation Function (PACF)',
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX

# -------------------------------------------------------------------------
# Step 1: Time-Based Train / Test Split (e.g., Last 30 days for testing)
# -------------------------------------------------------------------------
TEST_DAYS = 30
train = df_daily.iloc[:-TEST_DAYS]
test = df_daily.iloc[-TEST_DAYS:]

# -------------------------------------------------------------------------
# Step 2: Build & Fit SARIMAX Model
# -------------------------------------------------------------------------
# Order: (p=1, d=0, q=1), Seasonal Order: (P=1, D=0, Q=1, s=30)
model = SARIMAX(
    train[Target_Col],
    order=(1, 0, 1),
    seasonal_order=(1, 0, 1, 30),
    enforce_stationarity=False,
    enforce_invertibility=False,
)

results = model.fit(disp=False)
print(results.summary())

# -------------------------------------------------------------------------
# Step 3: Out-of-Sample Forecast & Evaluation Plot
# -------------------------------------------------------------------------
forecast = results.get_forecast(steps=TEST_DAYS)
forecast_df = forecast.conf_int()
forecast_df['predictions'] = forecast.predicted_mean

plt.figure(figsize=(14, 6))
plt.plot(
    train.index[-90:], train[Target_Col].values[-90:], label='Train (Recent)'
)
plt.plot(test.index, test[Target_Col].values, label='Actual Test Sales')
plt.plot(
    test.index,
    forecast_df['predictions'].values,
    color='red',
    label='SARIMAX Forecast',
)
plt.fill_between(
    test.index,
    forecast_df.iloc[:, 0].values,
    forecast_df.iloc[:, 1].values,
    color='pink',
    alpha=0.3,
    label='95% CI',
)

plt.title('Daily Revenue Forecast vs Actual (SARIMAX baseline)')
plt.xlabel('Days')
plt.ylabel('Final Amount (in lakhs)')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error
from statsmodels.tsa.statespace.sarimax import SARIMAX

# -------------------------------------------------------------------------
# Step 1: Feature Engineering (Calendar Exogenous Variables)
# -------------------------------------------------------------------------
# Convert date column to datetime if not already set
df_daily[Date_Col] = pd.to_datetime(df_daily[Date_Col])

# Flag weekends (Saturday/Sunday) and month-end dates
df_daily['day_of_week'] = df_daily[Date_Col].dt.dayofweek
df_daily['is_month_end'] = df_daily[Date_Col].dt.is_month_end.astype(int)

df_daily = pd.get_dummies(
    df_daily, columns=['day_of_week'], drop_first=True, dtype=int
)

# Identify all exogenous feature columns
EXOG_COLS = [c for c in df_daily.columns if 'day_of_week_' in c] + [
    'is_month_end'
]

# -------------------------------------------------------------------------
# Step 2: Time-Based Train/Test Split (Hold out last 31 days)
# -------------------------------------------------------------------------
TEST_DAYS = 31

train = df_daily.iloc[:-TEST_DAYS].copy()
test = df_daily.iloc[-TEST_DAYS:].copy()

print(f"Training set date range: {train[Date_Col].min().date()} to {train[Date_Col].max().date()}")
print(f"Testing set date range:  {test[Date_Col].min().date()} to {test[Date_Col].max().date()}")

# -------------------------------------------------------------------------
# Step 3: Fit SARIMAX(1,0,1)(1,0,1,31) with Exogenous Features
# -------------------------------------------------------------------------
model = SARIMAX(
    train[Target_Col],
    exog=train[EXOG_COLS],
    order=(1, 0, 1),
    seasonal_order=(1, 0, 1, 31),
    enforce_stationarity=False,
    enforce_invertibility=False,
)

model_fit = model.fit(disp=False)

# Display statistical summary
print(model_fit.summary())

# -------------------------------------------------------------------------
# Step 4: Generate Out-of-Sample Predictions Passing Test Exogenous Data
# -------------------------------------------------------------------------
forecast_results = model_fit.get_forecast(
    steps=TEST_DAYS, exog=test[EXOG_COLS]
)
forecast_df = forecast_results.conf_int()
forecast_df['Predictions'] = forecast_results.predicted_mean

# Append predictions to the test dataset
test['Predicted_Revenue'] = forecast_df['Predictions'].values

# Calculate Correct Error Metrics (wMAPE avoids divide-by-zero on 0-sales days)
rmse = np.sqrt(mean_squared_error(test[Target_Col], test['Predicted_Revenue']))
wmape = (
    np.sum(np.abs(test[Target_Col] - test['Predicted_Revenue']))
    / np.sum(test[Target_Col])
) * 100

print('=== Flexible Calendar Model Performance ===')
print(f'RMSE:  {rmse:.2f} lakhs')
print(f'wMAPE: {wmape:.2f}%')

# -------------------------------------------------------------------------
# Step 5: Visualize Predictions vs Actual Revenue
# -------------------------------------------------------------------------
plt.figure(figsize=(15, 6))

# Plot recent historical training data (last 90 days of training)
plt.plot(
    train[Date_Col].tail(90),
    train[Target_Col].tail(90),
    label='Historical Revenue (Train)',
    color='blue',
)

# Plot actual test revenue
plt.plot(
    test[Date_Col],
    test[Target_Col],
    label='Actual Revenue (Test)',
    color='black',
    linewidth=2,
)

# Plot model predictions
plt.plot(
    test[Date_Col],
    test['Predicted_Revenue'],
    label='SARIMAX Forecast (with Exog)',
    color='red',
    linestyle='--',
    linewidth=2,
)

# Plot 95% Confidence Interval
plt.fill_between(
    test[Date_Col],
    forecast_df.iloc[:, 0].values,
    forecast_df.iloc[:, 1].values,
    color='pink',
    alpha=0.3,
    label='95% Confidence Interval',
)

plt.title('Daily Revenue Forecast: Actual vs SARIMAX(1,0,1)(1,0,1,31) + Exog', fontsize=14)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Final Amount (in lakhs)', fontsize=12)
plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# -------------------------------------------------------------------------
# Step 6: Model Residual Diagnostics
# -------------------------------------------------------------------------
model_fit.plot_diagnostics(figsize=(15, 10))
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX

# -------------------------------------------------------------------------
# Step 1: Slice Historical Data up to Today (Sep 23)
# -------------------------------------------------------------------------
# Filter historical training data up to September 23
df_daily[Date_Col] = pd.to_datetime(df_daily[Date_Col])
train_history = df_daily[df_daily[Date_Col] <= '2026-09-23'].copy()

# Ensure day_of_week and is_month_end columns are present
EXOG_COLS = [c for c in train_history.columns if 'day_of_week_' in c] + [
    'is_month_end'
]

# -------------------------------------------------------------------------
# Step 2: Fit SARIMAX Model on Complete Updated History
# -------------------------------------------------------------------------
model = SARIMAX(
    train_history[Target_Col],
    exog=train_history[EXOG_COLS],
    order=(1, 0, 1),
    seasonal_order=(1, 0, 1, 31),
    enforce_stationarity=False,
    enforce_invertibility=False,
)

model_fit = model.fit(disp=False)

# -------------------------------------------------------------------------
# Step 3: Create Future Calendar DataFrame (Sep 24 to Sep 30 - 7 Days)
# -------------------------------------------------------------------------
future_dates = pd.date_range(start='2026-09-24', end='2026-09-30', freq='D')
future_df = pd.DataFrame({Date_Col: future_dates})

# Extract future exogenous features matching training structure
future_df['day_of_week'] = future_df[Date_Col].dt.dayofweek
future_df['is_month_end'] = future_df[Date_Col].dt.is_month_end.astype(int)

# One-hot encode day_of_week to match EXOG_COLS exactly
future_df = pd.get_dummies(
    future_df, columns=['day_of_week'], drop_first=True, dtype=int
)

# Align future columns with model training columns (filling missing dummy columns with 0)
for col in EXOG_COLS:
    if col not in future_df.columns:
        future_df[col] = 0

future_exog = future_df[EXOG_COLS]

# -------------------------------------------------------------------------
# Step 4: Generate Future 7-Day Revenue Forecast
# -------------------------------------------------------------------------
forecast = model_fit.get_forecast(steps=len(future_df), exog=future_exog)
forecast_summary = forecast.conf_int()
forecast_summary['Forecast_Revenue_Lakhs'] = forecast.predicted_mean.values
forecast_summary['Date'] = future_dates.values

# Clean table output
forecast_table = forecast_summary[
    [
        'Date',
        'Forecast_Revenue_Lakhs',
        'lower Final Amount (in lakhs)',
        'upper Final Amount (in lakhs)',
    ]
]
print('=== Forecasted Sales: Sep 24 to Sep 30 ===')
print(forecast_table.to_string(index=False))

# Calculate predicted total September revenue (Actuals till Sep 23 + Forecast Sep 24-30)
actual_sep_so_far = train_history[
    train_history[Date_Col].dt.month == 9
][Target_Col].sum()
projected_sep_remaining = forecast_summary['Forecast_Revenue_Lakhs'].sum()
total_projected_sep_revenue = actual_sep_so_far + projected_sep_remaining

print(f'\nActual Revenue (Sep 1 to Sep 23):     {actual_sep_so_far:.2f} lakhs')
print(f'Forecasted Revenue (Sep 24 to Sep 30): {projected_sep_remaining:.2f} lakhs')
print(f'Projected Total September Revenue:    {total_projected_sep_revenue:.2f} lakhs')

# -------------------------------------------------------------------------
# Step 5: Plot Historicals + Future Forecast
# -------------------------------------------------------------------------
plt.figure(figsize=(14, 6))

# Plot recent actual history (September 1 to 23)
sep_actuals = train_history[train_history[Date_Col] >= '2026-09-01']
plt.plot(
    sep_actuals[Date_Col],
    sep_actuals[Target_Col],
    label='Actual Revenue (Sep 1–23)',
    color='black',
    marker='o',
)

# Plot future predictions (September 24 to 30)
plt.plot(
    future_df[Date_Col],
    forecast_summary['Forecast_Revenue_Lakhs'],
    label='Predicted Revenue (Sep 24–30)',
    color='red',
    linestyle='--',
    marker='o',
)

# Plot confidence intervals for future days
plt.fill_between(
    future_df[Date_Col],
    forecast_summary.iloc[:, 0],
    forecast_summary.iloc[:, 1],
    color='pink',
    alpha=0.4,
    label='95% Confidence Interval',
)

plt.title('September Month-End Sales Forecast (Sep 24 to Sep 30)', fontsize=14)
plt.xlabel('Date')
plt.ylabel('Final Amount (in lakhs)')
plt.legend()
plt.label()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX

# -------------------------------------------------------------------------
# Step 1: Filter Training History up to September 23, 2026
# -------------------------------------------------------------------------
df_daily[Date_Col] = pd.to_datetime(df_daily[Date_Col])
train_history = df_daily[df_daily[Date_Col] <= '2026-09-23'].copy()

# Ensure exogenous dummy variables exist
EXOG_COLS = [c for c in train_history.columns if 'day_of_week_' in c] + [
    'is_month_end'
]

# -------------------------------------------------------------------------
# Step 2: Fit SARIMAX Model on Data up to Sep 23
# -------------------------------------------------------------------------
model = SARIMAX(
    train_history[Target_Col],
    exog=train_history[EXOG_COLS],
    order=(1, 0, 1),
    seasonal_order=(1, 0, 1, 31),
    enforce_stationarity=False,
    enforce_invertibility=False,
)

model_fit = model.fit(disp=False)

# -------------------------------------------------------------------------
# Step 3: Create Calendar Exogenous Features for Sep 24 to Sep 30 (7 Days)
# -------------------------------------------------------------------------
future_dates = pd.date_range(start='2026-09-24', end='2026-09-30', freq='D')
future_df = pd.DataFrame({Date_Col: future_dates})

future_df['day_of_week'] = future_df[Date_Col].dt.dayofweek
future_df['is_month_end'] = future_df[Date_Col].dt.is_month_end.astype(int)

# One-hot encode day_of_week to match training EXOG_COLS
future_df = pd.get_dummies(
    future_df, columns=['day_of_week'], drop_first=True, dtype=int
)

# Align columns with training data (fill missing dummies with 0)
for col in EXOG_COLS:
    if col not in future_df.columns:
        future_df[col] = 0

future_exog = future_df[EXOG_COLS]

# -------------------------------------------------------------------------
# Step 4: Generate Out-of-Sample Predictions
# -------------------------------------------------------------------------
forecast = model_fit.get_forecast(steps=len(future_df), exog=future_exog)
forecast_summary = forecast.conf_int()
forecast_summary['Forecast_Revenue_Lakhs'] = forecast.predicted_mean.values
forecast_summary['Date'] = future_dates

# Format display table
forecast_table = forecast_summary[
    [
        'Date',
        'Forecast_Revenue_Lakhs',
        'lower Final Amount (in lakhs)',
        'upper Final Amount (in lakhs)',
    ]
].copy()
forecast_table.columns = [
    'Date',
    'Predicted Revenue (Lakhs)',
    'Lower 95% CI',
    'Upper 95% CI',
]

print('=== September 24 to September 30 Revenue Forecast ===')
print(forecast_table.to_string(index=False))

# -------------------------------------------------------------------------
# Step 5: Full September Month Total Revenue Calculation
# -------------------------------------------------------------------------
actual_sep_so_far = train_history[
    (train_history[Date_Col].dt.month == 9)
    & (train_history[Date_Col].dt.year == 2026)
][Target_Col].sum()

projected_sep_remaining = forecast_summary['Forecast_Revenue_Lakhs'].sum()
total_projected_sep_revenue = actual_sep_so_far + projected_sep_remaining

print('\n=== September Month Summary Projection ===')
print(f'Actual Revenue Recorded (Sep 01 - Sep 23): {actual_sep_so_far:.2f} lakhs')
print(
    f'Forecasted Revenue (Sep 24 - Sep 30):      {projected_sep_remaining:.2f}'
    ' lakhs'
)
print(
    '---------------------------------------------------------------------'
)
print(
    'Total Projected September Closing Revenue: '
    f' {total_projected_sep_revenue:.2f} lakhs'
)

# -------------------------------------------------------------------------
# Step 6: Plot September Historical Sales + Future Forecast
# -------------------------------------------------------------------------
plt.figure(figsize=(16, 7))

sep_actuals = train_history[
    (train_history[Date_Col] >= '2026-09-01')
    & (train_history[Date_Col] <= '2026-09-23')
]

# Plot actual September 1–23 historicals
plt.plot(
    sep_actuals[Date_Col],
    sep_actuals[Target_Col],
    label='Actual Recorded Revenue (Sep 1–23)',
    color='black',
    marker='o',
    linewidth=2,
)

# Add Data Labels for Actuals
for x, y in zip(sep_actuals[Date_Col], sep_actuals[Target_Col]):
    plt.annotate(
        f'{y:.1f}',
        (x, y),
        textcoords='offset points',
        xytext=(0, 8),
        ha='center',
        fontsize=8,
        color='black',
        fontweight='bold',
    )

# Plot September 24–30 forecast
plt.plot(
    future_df[Date_Col],
    forecast_summary['Forecast_Revenue_Lakhs'],
    label='Predicted Revenue (Sep 24–30)',
    color='red',
    linestyle='--',
    marker='o',
    linewidth=2,
)

# Add Data Labels for Predictions
for x, y in zip(
    future_df[Date_Col], forecast_summary['Forecast_Revenue_Lakhs']
):
    plt.annotate(
        f'{y:.1f}',
        (x, y),
        textcoords='offset points',
        xytext=(0, 8),
        ha='center',
        fontsize=8,
        color='red',
        fontweight='bold',
    )

# Plot 95% Confidence Interval
plt.fill_between(
    future_df[Date_Col],
    forecast_summary.iloc[:, 0],
    forecast_summary.iloc[:, 1],
    color='pink',
    alpha=0.4,
    label='95% Confidence Interval',
)

plt.title(
    'Daily Revenue Forecast: September 24 to September 30, 2026', fontsize=14
)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Final Amount (in lakhs)', fontsize=12)

# Adjust y-axis top margin so data labels don't get cut off at the top
y_max = max(
    sep_actuals[Target_Col].max(),
    forecast_summary['Forecast_Revenue_Lakhs'].max(),
)
plt.ylim(top=y_max * 1.15)

plt.legend(loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX

# -------------------------------------------------------------------------
# Step 1: Filter Training History up to September 23, 2026
# -------------------------------------------------------------------------
df_daily[Date_Col] = pd.to_datetime(df_daily[Date_Col])
train_history = df_daily[df_daily[Date_Col] <= '2026-09-23'].copy()

# Ensure exogenous dummy variables exist
EXOG_COLS = [c for c in train_history.columns if 'day_of_week_' in c] + [
    'is_month_end'
]

# -------------------------------------------------------------------------
# Step 2: Fit SARIMAX Model on Data up to Sep 23
# -------------------------------------------------------------------------
model = SARIMAX(
    train_history[Target_Col],
    exog=train_history[EXOG_COLS],
    order=(1, 0, 1),
    seasonal_order=(1, 0, 1, 31),
    enforce_stationarity=False,
    enforce_invertibility=False,
)

model_fit = model.fit(method='powell', disp=False)

# -------------------------------------------------------------------------
# Step 3: Generate Forecast for Sep 24 to Sep 30 (7 Days)
# -------------------------------------------------------------------------
future_dates = pd.date_range(start='2026-09-24', end='2026-09-30', freq='D')
future_df = pd.DataFrame({Date_Col: future_dates})

future_df['day_of_week'] = future_df[Date_Col].dt.dayofweek
future_df['is_month_end'] = future_df[Date_Col].dt.is_month_end.astype(int)

future_df = pd.get_dummies(
    future_df, columns=['day_of_week'], drop_first=True, dtype=int
)

for col in EXOG_COLS:
    if col not in future_df.columns:
        future_df[col] = 0

future_exog = future_df[EXOG_COLS]

forecast = model_fit.get_forecast(steps=len(future_df), exog=future_exog)
future_df['Predicted_Revenue'] = forecast.predicted_mean.values

# -------------------------------------------------------------------------
# Step 4: Combine Daily Actuals & Forecasts into One Sep 2026 Series
# -------------------------------------------------------------------------
# Extract Sep 1 to Sep 23 actuals
sep_actuals = train_history[
    (train_history[Date_Col] >= '2026-09-01')
    & (train_history[Date_Col] <= '2026-09-23')
][[Date_Col, Target_Col]].copy()
sep_actuals.columns = [Date_Col, 'Revenue']
sep_actuals['Type'] = 'Actual'

# Extract Sep 24 to Sep 30 forecasts
sep_forecasts = future_df[[Date_Col, 'Predicted_Revenue']].copy()
sep_forecasts.columns = [Date_Col, 'Revenue']
sep_forecasts['Type'] = 'Forecast'

# Combine into single daily dataframe for September 2026
sep_combined = pd.concat([sep_actuals, sep_forecasts], ignore_index=True)

# Define 7-Day Weekly Buckets for September 2026
def assign_sep_week(date):
    day = date.day
    if day <= 7:
        return 'Week 1 (Sep 01-07)'
    elif day <= 14:
        return 'Week 2 (Sep 08-14)'
    elif day <= 21:
        return 'Week 3 (Sep 15-21)'
    else:
        return 'Week 4 (Sep 22-30)'

sep_combined['Week_Bucket'] = sep_combined[Date_Col].apply(assign_sep_week)

# -------------------------------------------------------------------------
# Step 5: Calculate Weekly Breakdown Table
# -------------------------------------------------------------------------
weekly_summary = []

for week, group in sep_combined.groupby('Week_Bucket', sort=False):
    actual_sum = group[group['Type'] == 'Actual']['Revenue'].sum()
    forecast_sum = group[group['Type'] == 'Forecast']['Revenue'].sum()
    total_sum = group['Revenue'].sum()
    
    weekly_summary.append({
        'Week': week,
        'Actual Revenue (Lakhs)': round(actual_sum, 2),
        'Forecast Revenue (Lakhs)': round(forecast_sum, 2),
        'Total Weekly Revenue (Lakhs)': round(total_sum, 2)
    })

weekly_df = pd.DataFrame(weekly_summary)

print('=== Weekly Revenue Breakdown: September 2026 ===')
print(weekly_df.to_string(index=False))

# -------------------------------------------------------------------------
# Step 6: Plot Weekly Bar Chart with Data Labels
# -------------------------------------------------------------------------
plt.figure(figsize=(12, 6))

x = np.arange(len(weekly_df['Week']))
width = 0.35

fig, ax = plt.subplots(figsize=(12, 6))

rects1 = ax.bar(
    x - width/2, 
    weekly_df['Actual Revenue (Lakhs)'], 
    width, 
    label='Actual Revenue', 
    color='black'
)
rects2 = ax.bar(
    x + width/2, 
    weekly_df['Forecast Revenue (Lakhs)'], 
    width, 
    label='Forecast Revenue', 
    color='red'
)

# Add Data Labels on Bars
def add_bar_labels(rects, color):
    for rect in rects:
        height = rect.get_height()
        if height > 0:
            ax.annotate(
                f'{height:.1f}',
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 5),
                textcoords="offset points",
                ha='center', 
                va='bottom',
                fontsize=10,
                fontweight='bold',
                color=color
            )

add_bar_labels(rects1, 'black')
add_bar_labels(rects2, 'red')

ax.set_title('Weekly Revenue Breakdown: September 2026 (Actual vs Forecast)', fontsize=14)
ax.set_ylabel('Revenue (in lakhs)', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(weekly_df['Week'], fontsize=11)
ax.set_ylim(top=max(weekly_df['Total Weekly Revenue (Lakhs)']) * 1.2)
ax.legend(loc='upper left')
ax.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX

# -------------------------------------------------------------------------
# Step 1: Filter Training History up to September 23, 2026
# -------------------------------------------------------------------------
df_daily[Date_Col] = pd.to_datetime(df_daily[Date_Col])
train_history = df_daily[df_daily[Date_Col] <= '2026-09-23'].copy()

# Exogenous dummy variables
EXOG_COLS = [c for c in train_history.columns if 'day_of_week_' in c] + [
    'is_month_end'
]

# -------------------------------------------------------------------------
# Step 2: Fit SARIMAX Model on Data up to Sep 23
# -------------------------------------------------------------------------
model = SARIMAX(
    train_history[Target_Col],
    exog=train_history[EXOG_COLS],
    order=(1, 0, 1),
    seasonal_order=(1, 0, 1, 31),
    enforce_stationarity=False,
    enforce_invertibility=False,
)

model_fit = model.fit(method='powell', disp=False)

# -------------------------------------------------------------------------
# Step 3: In-Sample Predictions (Sep 1–23) + Out-of-Sample Forecast (Sep 24–30)
# -------------------------------------------------------------------------
# 3a. In-sample predictions for Sep 1 to Sep 23
in_sample_preds = model_fit.predict(
    start=train_history[train_history[Date_Col] == '2026-09-01'].index[0],
    end=train_history[train_history[Date_Col] == '2026-09-23'].index[-1],
    exog=train_history[train_history[Date_Col] >= '2026-09-01'][EXOG_COLS],
)

sep_actuals = train_history[
    (train_history[Date_Col] >= '2026-09-01')
    & (train_history[Date_Col] <= '2026-09-23')
][[Date_Col, Target_Col]].copy()
sep_actuals['Predicted_Revenue'] = in_sample_preds.values

# 3b. Out-of-sample forecast for Sep 24 to Sep 30
future_dates = pd.date_range(start='2026-09-24', end='2026-09-30', freq='D')
future_df = pd.DataFrame({Date_Col: future_dates})

future_df['day_of_week'] = future_df[Date_Col].dt.dayofweek
future_df['is_month_end'] = future_df[Date_Col].dt.is_month_end.astype(int)

future_df = pd.get_dummies(
    future_df, columns=['day_of_week'], drop_first=True, dtype=int
)

for col in EXOG_COLS:
    if col not in future_df.columns:
        future_df[col] = 0

future_exog = future_df[EXOG_COLS]
forecast = model_fit.get_forecast(steps=len(future_df), exog=future_exog)

future_df[Target_Col] = 0.0  # No actual revenue recorded yet for future dates
future_df['Predicted_Revenue'] = forecast.predicted_mean.values

# Combine Sep 1-23 and Sep 24-30 into one full September DataFrame
sep_full = pd.concat(
    [sep_actuals[[Date_Col, Target_Col, 'Predicted_Revenue']], future_df[[Date_Col, Target_Col, 'Predicted_Revenue']]],
    ignore_index=True,
)

# -------------------------------------------------------------------------
# Step 4: Define Weekly Buckets & Aggregate
# -------------------------------------------------------------------------
def assign_sep_week(date):
    day = date.day
    if day <= 7:
        return 'Week 1 (Sep 01-07)'
    elif day <= 14:
        return 'Week 2 (Sep 08-14)'
    elif day <= 21:
        return 'Week 3 (Sep 15-21)'
    else:
        return 'Week 4 (Sep 22-30)'

sep_full['Week_Bucket'] = sep_full[Date_Col].apply(assign_sep_week)

weekly_summary = []
for week, group in sep_full.groupby('Week_Bucket', sort=False):
    actual_sum = group[Target_Col].sum()
    forecast_sum = group['Predicted_Revenue'].sum()
    diff = forecast_sum - actual_sum if actual_sum > 0 else np.nan

    weekly_summary.append({
        'Week': week,
        'Actual Revenue (Lakhs)': round(actual_sum, 2),
        'Model Forecast (Lakhs)': round(forecast_sum, 2),
        'Variance (Lakhs)': round(diff, 2) if not np.isnan(diff) else 'N/A (Future)',
    })

weekly_df = pd.DataFrame(weekly_summary)

print('=== Complete September 2026 Weekly Actual vs Forecast Comparison ===')
print(weekly_df.to_string(index=False))

# -------------------------------------------------------------------------
# Step 5: Plot Side-by-Side Bar Chart for All Weeks
# -------------------------------------------------------------------------
x = np.arange(len(weekly_df['Week']))
width = 0.35

fig, ax = plt.subplots(figsize=(13, 6))

rects1 = ax.bar(
    x - width / 2,
    weekly_df['Actual Revenue (Lakhs)'],
    width,
    label='Actual Revenue',
    color='black',
)
rects2 = ax.bar(
    x + width / 2,
    weekly_df['Model Forecast (Lakhs)'],
    width,
    label='Model Forecast',
    color='crimson',
)

def add_bar_labels(rects, color):
    for rect in rects:
        height = rect.get_height()
        if height > 0:
            ax.annotate(
                f'{height:.1f}',
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 5),
                textcoords='offset points',
                ha='center',
                va='bottom',
                fontsize=10,
                fontweight='bold',
                color=color,
            )

add_bar_labels(rects1, 'black')
add_bar_labels(rects2, 'crimson')

ax.set_title(
    'September 2026: Weekly Actual vs Model Forecast Comparison',
    fontsize=14,
)
ax.set_ylabel('Revenue (in lakhs)', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(weekly_df['Week'], fontsize=11)
ax.set_ylim(top=max(weekly_df['Model Forecast (Lakhs)'].max(), weekly_df['Actual Revenue (Lakhs)'].max()) * 1.25)
ax.legend(loc='upper left')
ax.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
